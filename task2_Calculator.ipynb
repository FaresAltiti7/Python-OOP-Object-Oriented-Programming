```python
class Calculator:

    def add(self, a, b):
        return a + b

    def subtract(self, a, b):
        return a - b

    def multiply(self, a, b):
        return a * b

    def divideInt(self, a, b):
        return a // b

    def divideFloat(self, a, b):
        return a / b

    def modulus(self, a, b):
        return a % b


calc = Calculator()

print("add =", calc.add(15, 10))
print("subtract=", calc.subtract(20, 8))
print("multiply", calc.multiply(6, 7))
print("divideint", calc.divideInt(7, 2))
print("dividefloat", calc.divideFloat(17, 5))
print("mpdulus", calc.modulus(10, 2))
```
